```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("Environment Ready!")
```

```python
# Raw data dictionary
raw_data = {
    'State': ['Lagos', 'Ogun', 'Oyo', 'Kano', 'Kaduna', 'Katsina', 'Enugu', 'Abia', 'Anambra', 'Bauchi', 'Borno'],
    'Budget_Pct': [12.5, 11.0, 13.1, 26.0, 24.5, 22.1, 15.2, 14.5, 16.0, 19.5, 22.0],
    'Pass_Rate': [82, 80, 79, 55, 58, 52, 78, 81, 85, 48, 45]
}

# Mapping Dictionary
zones = {
    'Lagos': 'South West', 'Ogun': 'South West', 'Oyo': 'South West',
    'Kano': 'North West', 'Kaduna': 'North West', 'Katsina': 'North West',
    'Enugu': 'South East', 'Abia': 'South East', 'Anambra': 'South East',
    'Bauchi': 'North East', 'Borno': 'North East'
}

# Create DataFrame and Map Zones
df = pd.DataFrame(raw_data)
df['Zone'] = df['State'].map(zones)
df
```

```python
# Calculating the average pass rate per zone
regional_analysis = df.groupby('Zone')['Pass_Rate'].mean().sort_values(ascending=False)
print("Average Pass Rate by Zone:")
print(regional_analysis)
```

```python
# Set the visual style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Create the Bar Chart
ax = sns.barplot(x=regional_analysis.index, y=regional_analysis.values, palette="viridis")

# Add titles and labels
plt.title('Average WAEC Pass Rate by Geopolitical Zone (2024)', fontsize=15)
plt.xlabel('Geopolitical Zone', fontsize=12)
plt.ylabel('Average Pass Rate (%)', fontsize=12)
plt.ylim(0, 100) # Keep the scale 0-100 for truthfulness

# Show the plot
plt.show()
```
