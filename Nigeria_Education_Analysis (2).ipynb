```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("Environment Ready!")
```

```python
# Raw data dictionary
raw_data = {
    'State': ['Lagos', 'Ogun', 'Oyo', 'Kano', 'Kaduna', 'Katsina', 'Enugu', 'Abia', 'Anambra', 'Bauchi', 'Borno'],
    'Budget_Pct': [12.5, 11.0, 13.1, 26.0, 24.5, 22.1, 15.2, 14.5, 16.0, 19.5, 22.0],
    'Pass_Rate': [82, 80, 79, 55, 58, 52, 78, 81, 85, 48, 45]
}

# Mapping Dictionary
zones = {
    'Lagos': 'South West', 'Ogun': 'South West', 'Oyo': 'South West',
    'Kano': 'North West', 'Kaduna': 'North West', 'Katsina': 'North West',
    'Enugu': 'South East', 'Abia': 'South East', 'Anambra': 'South East',
    'Bauchi': 'North East', 'Borno': 'North East'
}

# Create DataFrame and Map Zones
df = pd.DataFrame(raw_data)
df['Zone'] = df['State'].map(zones)
df
```

```python
# Calculating the average pass rate per zone
regional_analysis = df.groupby('Zone')['Pass_Rate'].mean().sort_values(ascending=False)
print("Average Pass Rate by Zone:")
print(regional_analysis)
```

```python
# Set the visual style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Create the Bar Chart
ax = sns.barplot(x=regional_analysis.index, y=regional_analysis.values, palette="viridis")

# Add titles and labels
plt.title('Average WAEC Pass Rate by Geopolitical Zone (2024)', fontsize=15)
plt.xlabel('Geopolitical Zone', fontsize=12)
plt.ylabel('Average Pass Rate (%)', fontsize=12)
plt.ylim(0, 100) # Keep the scale 0-100 for truthfulness

# Show the plot
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Create a scatter plot
sns.scatterplot(data=df, x='Budget_Pct', y='Pass_Rate', hue='Zone', s=100)

# Add a "Trend Line" to see the overall direction
sns.regplot(data=df, x='Budget_Pct', y='Pass_Rate', scatter=False, color='red')

plt.title('Relationship: Education Budget % vs. Exam Pass Rate', fontsize=15)
plt.xlabel('State Education Budget (%)', fontsize=12)
plt.ylabel('WAEC Pass Rate (%)', fontsize=12)

plt.show()
```

📔 Conclusion & Summary of Findings
1. The Result: A Counter-Intuitive Trend 📉

The analysis revealed a surprising negative correlation between the percentage of the state budget allocated to education and the WAEC pass rates. While one might expect higher spending to lead to higher scores, the scatter plot and trend line showed that, in this specific dataset, states with the highest budget percentages often had lower performance than those with more modest allocations.

2. The Interpretation: Efficiency Over Volume 💡

This "Spending Paradox" suggests that the volume of money is not the primary driver of academic success. Instead, the results point toward spending efficiency. The South East zone, for example, maintained dominant pass rates despite not always having the highest budget percentages. This implies that factors such as existing infrastructure, teacher quality, community involvement, and historical literacy rates play a more significant role than the current year's budget figures.

3. The Recommendation: Beyond the Numbers 🚀

For policymakers, this data suggests that simply increasing the education budget may not be enough to improve student outcomes. Future research should investigate how funds are being spent—focusing on teacher training, learning materials, and rural school support—rather than just how much is being allocated.
